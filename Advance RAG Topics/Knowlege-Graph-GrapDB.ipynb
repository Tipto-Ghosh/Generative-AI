```python
import os 
from dotenv import load_dotenv
load_dotenv()


# graph db related constants
NEO4J_URI = os.getenv("NEO4J_URI")
NEO4J_USERNAME = os.getenv("NEO4J_USERNAME")
NEO4J_PASSWORD = os.getenv("NEO4J_PASSWORD")
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
os.environ["NEO4J_URI"] = NEO4J_URI
os.environ["NEO4J_USERNAME"] = NEO4J_USERNAME
os.environ["NEO4J_PASSWORD"] = NEO4J_PASSWORD
```

```python
from langchain_neo4j import Neo4jGraph

graph = Neo4jGraph(
    url = NEO4J_URI,
    username = NEO4J_USERNAME,
    password = NEO4J_PASSWORD
)
```

```python
from langchain_groq.chat_models import ChatGroq

chat_model = ChatGroq(
    api_key = groq_api_key,
    model = "llama-3.1-8b-instant" 
)
```

```python
chat_model
```

```python
from langchain_core.documents import Document

text = """Elon Reeve Musk is a visionary entrepreneur born on June 28, 1971, in Pretoria, 
South Africa, to a Canadian mother and a South African father. 
A self-taught programmer from age 12, he later moved to Canada and then the United States 
to earn degrees in Physics and Economics from the University of Pennsylvania. 
He became a multi-millionaire in his 20s after selling his first company, Zip2, 
and co-founding X.com, which eventually became PayPal. 
Today, he is the CEO and lead designer of SpaceX, where he pioneered reusable rockets, 
and the CEO of Tesla, which revolutionized the electric vehicle industry. 
Musk also leads several other cutting-edge ventures, 
including Neuralink (brain-chip interfaces), The Boring Company (tunneling), and 
the social media platform X (formerly Twitter).
"""

document = [Document(page_content = text)]
```

```python
document
```

```python
from langchain_experimental.graph_transformers import LLMGraphTransformer

llm_transformer = LLMGraphTransformer(
    llm = chat_model
)
```

```python
graph_documents = llm_transformer.convert_to_graph_documents(documents = document)
```
